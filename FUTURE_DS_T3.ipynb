```python

# Step 1: Import required libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python

# Install UCI dataset library

!pip install ucimlrepo

# Import dataset function
from ucimlrepo import fetch_ucirepo

# Fetch Bank Marketing dataset
bank_marketing = fetch_ucirepo(id=222)

# Combine features and target
df = pd.concat(
    [bank_marketing.data.features,
     bank_marketing.data.targets],
    axis=1
)

# Display first 5 rows
df.head()
```

```python

# Number of rows and columns
print("Dataset Shape:", df.shape)

# Column names
print("\nColumn Names:")
print(df.columns.tolist())

# Dataset information
print("\nDataset Information:")
df.info()

# Statistical summary
df.describe(include="all")
```

```python

# Create a copy of the original dataset

df_clean = df.copy()

print("Original dataset:", df.shape)
print("Working dataset:", df_clean.shape)
```

```python

# Check missing values in each column

print("Missing Values:")
print(df_clean.isnull().sum())

# Total missing values
print("\nTotal Missing Values:",
      df_clean.isnull().sum().sum())
```

```python

# Count duplicate rows

print("Duplicate Rows:",
      df_clean.duplicated().sum())
```

```python

# Remove duplicate records

df_clean = df_clean.drop_duplicates()

print("Dataset after removing duplicates:")
print(df_clean.shape)
```

```python

# Display data types

print(df_clean.dtypes)
```

```python

print("FINAL DATASET SUMMARY")

print("Rows:", df_clean.shape[0])
print("Columns:", df_clean.shape[1])
print("Missing Values:", df_clean.isnull().sum().sum())
print("Duplicate Rows:", df_clean.duplicated().sum())

print("\nData Cleaning Completed!")
```

```python

# Count customers who subscribed and did not subscribe

subscription_count = df_clean['y'].value_counts()

print(subscription_count)

# Visualize subscription distribution

plt.figure(figsize=(7,5))

sns.countplot(data=df_clean, x='y')

plt.title('Customer Subscription Distribution')
plt.xlabel('Subscription Status')
plt.ylabel('Number of Customers')

plt.show()
```

```python

# Total customers contacted
total_customers = len(df_clean)

# Successful subscriptions
converted_customers = (df_clean['y'] == 'yes').sum()

# Conversion rate
conversion_rate = (converted_customers / total_customers) * 100

print("Total Customers:", total_customers)
print("Converted Customers:", converted_customers)
print("Overall Conversion Rate:", round(conversion_rate, 2), "%")
```

```python

# Calculate conversion rate by job

job_conversion = (
    df_clean.groupby('job')['y']
    .apply(lambda x: (x == 'yes').mean() * 100)
    .sort_values(ascending=False)
)

print(job_conversion)

# Create graph
plt.figure(figsize=(12,6))

job_conversion.plot(kind='bar')

plt.title('Conversion Rate by Job')
plt.xlabel('Job')
plt.ylabel('Conversion Rate (%)')
plt.xticks(rotation=45)

plt.show()
```

```python

# Conversion rate by contact method

channel_conversion = (
    df_clean.groupby('contact')['y']
    .apply(lambda x: (x == 'yes').mean() * 100)
)

print(channel_conversion)

plt.figure(figsize=(7,5))

channel_conversion.plot(kind='bar')

plt.title('Conversion Rate by Contact Channel')
plt.xlabel('Contact Channel')
plt.ylabel('Conversion Rate (%)')

plt.show()
```

```python

# Conversion rate by month

monthly_conversion = (
    df_clean.groupby('month')['y']
    .apply(lambda x: (x == 'yes').mean() * 100)
)

print(monthly_conversion)

plt.figure(figsize=(10,5))

monthly_conversion.plot(kind='bar')

plt.title('Monthly Marketing Conversion Rate')
plt.xlabel('Month')
plt.ylabel('Conversion Rate (%)')

plt.show()
```

```python

# Conversion rate by previous campaign outcome

previous_conversion = (
    df_clean.groupby('poutcome')['y']
    .apply(lambda x: (x == 'yes').mean() * 100)
)

print(previous_conversion)

plt.figure(figsize=(8,5))

previous_conversion.plot(kind='bar')

plt.title('Conversion by Previous Campaign Result')
plt.xlabel('Previous Campaign Result')
plt.ylabel('Conversion Rate (%)')

plt.show()
```

```python

# Marketing Funnel Analysis

total_contacts = len(df_clean)

successful_conversions = (df_clean['y'] == 'yes').sum()

unsuccessful_conversions = (df_clean['y'] == 'no').sum()

# Conversion rate
conversion_rate = (
    successful_conversions / total_contacts
) * 100

# Drop-off rate
drop_off_rate = (
    unsuccessful_conversions / total_contacts
) * 100

print("MARKETING FUNNEL ANALYSIS")
print("-------------------------")

print("Total Contacts:", total_contacts)
print("Successful Conversions:", successful_conversions)
print("Unsuccessful Conversions:", unsuccessful_conversions)

print("Conversion Rate:", round(conversion_rate, 2), "%")
print("Drop-off Rate:", round(drop_off_rate, 2), "%")
```

```python

# Funnel stage analysis

funnel_data = pd.DataFrame({
    'Stage': ['Contacted Prospects',
              'Successful Conversions'],
    'Count': [total_contacts,
              successful_conversions]
})

print(funnel_data)
```

```python

plt.figure(figsize=(9, 5))

plt.bar(
    funnel_data['Stage'],
    funnel_data['Count']
)

plt.title('Marketing Conversion Funnel')
plt.xlabel('Funnel Stages')
plt.ylabel('Number of Customers')

for i, value in enumerate(funnel_data['Count']):
    plt.text(i, value, str(value), ha='center', va='bottom')

plt.show()
```

```python

# Conversion and drop-off comparison

labels = ['Converted', 'Not Converted']

values = [
    successful_conversions,
    unsuccessful_conversions
]

plt.figure(figsize=(7, 7))

plt.pie(
    values,
    labels=labels,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Marketing Conversion and Drop-off')

plt.show()
```

```python

print("BUSINESS INSIGHTS")
print("-----------------")

print(
    f"Out of {total_contacts} contacted prospects, "
    f"{successful_conversions} subscribed."
)

print(
    f"Overall conversion rate is "
    f"{conversion_rate:.2f}%."
)

print(
    f"Drop-off rate is "
    f"{drop_off_rate:.2f}%."
)

if conversion_rate < 10:
    print("Recommendation: Improve campaign targeting and customer engagement.")
else:
    print("Recommendation: Study successful campaigns and scale effective strategies.")
```

```python

# Analyze marketing channel performance

channel_analysis = df_clean.groupby('contact').agg(
    Total_Contacts=('y', 'count'),
    Successful_Conversions=('y',
        lambda x: (x == 'yes').sum())
)

channel_analysis['Conversion_Rate'] = (
    channel_analysis['Successful_Conversions'] /
    channel_analysis['Total_Contacts'] * 100
)

print(channel_analysis.sort_values(
    by='Conversion_Rate', ascending=False
))
```

```python

plt.figure(figsize=(8,5))

sns.barplot(
    data=channel_analysis.reset_index(),
    x='contact',
    y='Conversion_Rate'
)

plt.title('Marketing Channel Conversion Performance')
plt.xlabel('Contact Channel')
plt.ylabel('Conversion Rate (%)')
plt.show()
```

```python

# Group campaign contact frequency

df_clean['campaign_group'] = pd.cut(
    df_clean['campaign'],
    bins=[0, 1, 3, 5, 10, float('inf')],
    labels=['1 Contact', '2-3 Contacts',
            '4-5 Contacts', '6-10 Contacts',
            'Above 10 Contacts']
)

campaign_analysis = df_clean.groupby(
    'campaign_group', observed=False
).agg(
    Total_Contacts=('y', 'count'),
    Conversions=('y', lambda x: (x == 'yes').sum())
)

campaign_analysis['Conversion_Rate'] = (
    campaign_analysis['Conversions'] /
    campaign_analysis['Total_Contacts'] * 100
)

print(campaign_analysis)
```

```python

plt.figure(figsize=(10,5))

campaign_analysis['Conversion_Rate'].plot(
    kind='bar'
)

plt.title('Conversion Rate by Campaign Contact Frequency')
plt.xlabel('Number of Contacts')
plt.ylabel('Conversion Rate (%)')
plt.xticks(rotation=30)

plt.show()
```

```python

# Arrange months in calendar order

month_order = [
    'jan', 'feb', 'mar', 'apr', 'may', 'jun',
    'jul', 'aug', 'sep', 'oct', 'nov', 'dec'
]

monthly_analysis = df_clean.groupby('month').agg(
    Total_Contacts=('y', 'count'),
    Conversions=('y', lambda x: (x == 'yes').sum())
)

monthly_analysis['Conversion_Rate'] = (
    monthly_analysis['Conversions'] /
    monthly_analysis['Total_Contacts'] * 100
)

monthly_analysis = monthly_analysis.reindex(month_order)

print(monthly_analysis)
```

```python

plt.figure(figsize=(11,5))

monthly_analysis['Conversion_Rate'].plot(
    marker='o',
    linewidth=2
)

plt.title('Monthly Marketing Campaign Performance')
plt.xlabel('Month')
plt.ylabel('Conversion Rate (%)')
plt.grid(True)

plt.show()
```

```python

# Step 8.1: Best Performing Marketing Channel

best_channel = channel_analysis['Conversion_Rate'].idxmax()
best_channel_rate = channel_analysis['Conversion_Rate'].max()

print("BEST MARKETING CHANNEL")
print("----------------------")
print("Channel:", best_channel)
print("Conversion Rate:", round(best_channel_rate, 2), "%")
```

```python

# Best performing month

best_month = monthly_analysis['Conversion_Rate'].idxmax()
best_month_rate = monthly_analysis['Conversion_Rate'].max()

print("BEST PERFORMING MONTH")
print("---------------------")
print("Month:", best_month)
print("Conversion Rate:", round(best_month_rate, 2), "%")
```

```python

# Best campaign contact frequency

best_campaign = campaign_analysis['Conversion_Rate'].idxmax()
best_campaign_rate = campaign_analysis['Conversion_Rate'].max()

print("BEST CAMPAIGN FREQUENCY")
print("-----------------------")
print("Contact Frequency:", best_campaign)
print("Conversion Rate:", round(best_campaign_rate, 2), "%")
```

```python

print("BUSINESS RECOMMENDATIONS")
print("========================")

print(f"""
1. Focus on the {best_channel} marketing channel,
   which has the highest observed conversion rate.

2. Study campaigns conducted in {best_month},
   which has the highest observed monthly conversion rate.

3. Analyze the {best_campaign} contact group
   to understand its successful marketing practices.

4. Improve customer targeting to reduce unsuccessful
   marketing contacts.

5. Monitor conversion rates regularly to improve
   future marketing campaigns.
""")
```

```python

# Create final insights table

insights = pd.DataFrame({
    'Category': [
        'Best Marketing Channel',
        'Best Performing Month',
        'Best Campaign Frequency'
    ],
    'Result': [
        str(best_channel),
        str(best_month),
        str(best_campaign)
    ],
    'Conversion Rate (%)': [
        round(best_channel_rate, 2),
        round(best_month_rate, 2),
        round(best_campaign_rate, 2)
    ]
})

print(insights)
```

```python

# Step 9.1: Create KPI Summary

kpi_data = {
    'Metric': [
        'Total Contacts',
        'Successful Conversions',
        'Unsuccessful Contacts',
        'Overall Conversion Rate (%)',
        'Overall Drop-off Rate (%)'
    ],
    'Value': [
        total_contacts,
        successful_conversions,
        unsuccessful_conversions,
        round(conversion_rate, 2),
        round(drop_off_rate, 2)
    ]
}

kpi_summary = pd.DataFrame(kpi_data)

print("MARKETING PERFORMANCE KPI SUMMARY")
print("=" * 40)
print(kpi_summary)
```

```python

# Step 9.2: Final Marketing Dashboard

fig, axes = plt.subplots(2, 2, figsize=(16, 11))

# 1. Subscription distribution
sns.countplot(data=df_clean, x='y', ax=axes[0, 0])
axes[0, 0].set_title('Customer Subscription Distribution')
axes[0, 0].set_xlabel('Subscription')
axes[0, 0].set_ylabel('Number of Customers')

# 2. Channel performance
channel_analysis['Conversion_Rate'].plot(
    kind='bar', ax=axes[0, 1]
)
axes[0, 1].set_title('Conversion by Marketing Channel')
axes[0, 1].set_xlabel('Contact Channel')
axes[0, 1].set_ylabel('Conversion Rate (%)')

# 3. Monthly performance
monthly_analysis['Conversion_Rate'].plot(
    kind='bar', ax=axes[1, 0]
)
axes[1, 0].set_title('Monthly Conversion Performance')
axes[1, 0].set_xlabel('Month')
axes[1, 0].set_ylabel('Conversion Rate (%)')

# 4. Campaign frequency
campaign_analysis['Conversion_Rate'].plot(
    kind='bar', ax=axes[1, 1]
)
axes[1, 1].set_title('Conversion by Campaign Frequency')
axes[1, 1].set_xlabel('Campaign Contacts')
axes[1, 1].set_ylabel('Conversion Rate (%)')

plt.suptitle(
    'Marketing Funnel & Conversion Performance Dashboard',
    fontsize=17
)

plt.tight_layout()
plt.show()
```
